```python
import pandas as pd
import sys

sys.path.append("..")

from src.lexicon_absa2_opinion_to_sentiment import LexiconOpinionFirstABSA

df = pd.read_csv("../data/Yelp Restaurant Reviews.csv")

num_rows = len(df)
print("Number of rows:", num_rows)
print("Dataset shape:", df.shape)
print("Columns:", df.columns.tolist())
print("\nSample reviews:")
print(df['Review Text'].head(10))

# (Run this once if the spaCy model isn't installed yet)
# !python -m spacy download en_core_web_sm

```

```python
analyzer = LexiconOpinionFirstABSA()

sample_reviews = df['Review Text'].sample(5)

for i, review in enumerate(sample_reviews, 1):
    results = analyzer.analyze(review)
    print(f"\nRandom Review {i}: {review}")
    for r in results:
        print("   ", r)

all_results = []
for review in sample_reviews:
    results = analyzer.analyze(review)
    all_results.append({
        "review": review,
        "aspects": [r.aspect for r in results],
        "sentiments": [r.sentiment for r in results],
        "confidences": [r.confidence for r in results]
    })

results_df = pd.DataFrame(all_results)
print("\nResults DataFrame:")
display(results_df)

all_aspects = [aspect for r in all_results for aspect in r['aspects']]
aspects_series = pd.Series(all_aspects)
top_aspects = aspects_series.value_counts().head(20)
print("\nTop 20 aspects found:")
print(top_aspects)
```
