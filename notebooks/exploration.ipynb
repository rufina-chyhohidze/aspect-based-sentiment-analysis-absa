```python
import pandas as pd
import sys


sys.path.append("..")

from src.lexicon_absa import LexiconABSA

# Load dataset once
df = pd.read_csv("../data/Yelp Restaurant Reviews.csv")

print("Dataset shape:", df.shape)
print("Columns:", df.columns.tolist())
print("\nSample reviews:")
print(df['Review Text'].head(10))
```

```python
# (Run this once if the spaCy model isn't installed yet)
# !python -m spacy download en_core_web_sm

# Initialize analyzer
analyzer = LexiconABSA()

# Run analyzer on 5 random reviews
sample_reviews = df['Review Text'].sample(5)

for i, review in enumerate(sample_reviews, 1):
    results = analyzer.analyze(review)
    print(f"\nRandom Review {i}: {review}")
    for r in results:
        print("   ", r)

# Collect results for those same random reviews
all_results = []
for review in sample_reviews:
    results = analyzer.analyze(review)
    all_results.append({
        "review": review,
        "aspects": [r.aspect for r in results],
        "sentiments": [r.sentiment for r in results]
    })

results_df = pd.DataFrame(all_results)
results_df
```
